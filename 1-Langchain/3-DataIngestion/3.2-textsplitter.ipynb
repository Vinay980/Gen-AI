#### Text Splitters

#### Text Splitting from Documents- RecursiveCharacter Text Splitters

This text splitter is the recommended one for generic text. It is parameterized by a list of characters. It tries to split on them in order until the chunks are small enough. The default list is ["\n\n","\n"," ",""]. This has the effect of trying to keep all paragraphs (and then sentences, and then words) together as long as possible, as those would generically seem to be the strongest semantically related pieces of text.

* How the text is split: by list or characters.
* how the chunk size is measured: by number of characters.

```python

## For pdf

from langchain_community.document_loaders import PyPDFLoader

loader = PyPDFLoader(
    "Discover how ChatGPT is trained!.pdf"
)

pdf_documents = loader.load()

print("Total pages:", len(pdf_documents))
print(pdf_documents[0].page_content)
```

#### How to recursively split text by characters

```python
from langchain_text_splitters import RecursiveCharacterTextSplitter

text_splitter = RecursiveCharacterTextSplitter(chunk_size = 500, chunk_overlap=50)
final_document = text_splitter.split_documents(pdf_documents)
final_document
```

```python
print(final_document[0])
print(final_document[1])
```

#### How to split by Character-Character Text Splitter

This is the simplest method. This splits based on a given character sequence, which defaults to "\n\n". Chuck length is measured by numbers or characters.

* 1. How the text is split: by single character seperator.
* 2. how the chunk size is measured: by number of characters.

```python
## text

from langchain_community.document_loaders import TextLoader

loader = TextLoader("speech.txt")

text_doc = loader.load()
```

```python
speech=''

with open('speech.txt') as f:
    speech = f.read()
    
text_splitter = RecursiveCharacterTextSplitter(chunk_size=100,chunk_overlap=20)
text = text_splitter.create_documents([speech])
print(text[0])
print(text[1])
print(text[2])

```
