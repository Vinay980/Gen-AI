#### Chroma

Chroma is a AI-native open_source vector database focused on developer productivity and happiness. Chroma is licensed under Apache 2.0.

```python
## building a sample vectordb
from langchain_chroma import Chroma
from langchain_community.document_loaders import TextLoader
from langchain_ollama import OllamaEmbeddings
from langchain_text_splitters import RecursiveCharacterTextSplitter
```

```python
loader = TextLoader("speech.txt")
data = loader.load()
data
```

```python
# Split
text_splitter = RecursiveCharacterTextSplitter(chunk_size = 500, chunk_overlap=0)
splits = text_splitter.split_documents(data)
```

```python
embedding = OllamaEmbeddings(model="nomic-embed-text")
vectordb = Chroma.from_documents(documents=splits,embedding=embedding)
vectordb
```

```python
### querying
query ="who's speech is this?" 
docs = vectordb.similarity_search(query)
docs[0].page_content
```

```python
## Saving to the disk
vectordb = Chroma.from_documents(documents=splits,embedding=embedding,persist_directory="./chroma.db")
```

```python
## load from disk
db2 = Chroma(persist_directory="./chroma.db",embedding_function=embedding)
docs = db2.similarity_search(query)
print(docs[0].page_content)
```

```python
## similarity Search with score
docs_and_score = vectordb.similarity_search_with_score(query)
docs_and_score
```

```python
## Retriever option
retriever = vectordb.as_retriever()
retriever.invoke(query)[0].page_content
```
